# Zbiór danych Titanic – analiza brakujących wartości

## Konfiguracja środowiska

```python
import pandas as pd
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import arff
```

```python
DATA_PATH = "data/Titanic.arff"

with open(DATA_PATH, "r", encoding="utf-8") as f:
    dataset = arff.load(f)

df = pd.DataFrame(
    dataset["data"],
    columns=[attribute[0] for attribute in dataset["attributes"]],
)
```

## 2. Analiza brakujących wartości

```python
missing_count = pd.isnull(df).sum()
missing_share = pd.isnull(df).mean()
has_missing = missing_count > 0

print(missing_count[has_missing])
print()
print(missing_share[has_missing])

cols_with_nan = missing_count[has_missing].index.tolist()
```

**Czy zbiór wymagał modyfikacji przed analizą braków?**

Braki danych występują w 7 z 14 kolumn. Wykryto je za pomocą funkcji `pd.isnull().sum()` oraz `pd.isnull().mean()`. Podczas wczytywania pliku `.arff` parser `arff.load()` automatycznie zamienił znaki zapytania (`?`), którymi w pliku źródłowym oznaczono brakujące wartości, na `None`, a Pandas rozpoznał je i zamienił na `np.nan`. Dzięki temu dodatkowa modyfikacja zbioru nie była potrzebna.

W przypadku wczytywania danych bezpośrednio z surowego pliku tekstowego konieczna byłaby ręczna konwersja `?` na `np.nan`. Można to zrobić np. za pomocą `df.replace("?", np.nan)` lub ustawiając w `pd.read_csv` parametr `na_values="?"`.

**Rozkład braków w cechach:**

- Duży udział braków (powyżej 60%): `body` (90,76%, 1188 braków), `cabin` (77,46%, 1014 braków) oraz `boat` (62,87%, 823 braki).
- Umiarkowany udział braków (20–45%): `home.dest` (43,09%, 564 braki) oraz `age` (20,09%, 263 braki).
- Niewielki udział braków (poniżej 1%): `embarked` (0,15%, 2 braki) oraz `fare` (0,08%, 1 brak).

## 3. Rodzaj braków danych


### 3.1. Zależność braków od zmiennej `survived`

```python
df["body"].isnull().groupby(df["survived"]).mean()
```

Odsetek braków w kolumnie `body` wśród ocalałych (`survived = 1`) wynosi dokładnie 100%, co oznacza, że żaden ocalały nie ma identyfikatora ciała. W grupie ofiar (`survived = 0`) odsetek braków wynosi 85%, co wskazuje, że ciał zdecydowanej większości zmarłych nigdy nie odnaleziono lub nie zidentyfikowano.

Brakujące dane w kolumnie `body` są typu **MNAR**, ponieważ stoi za tym oczywisty mechanizm przyczynowo-skutkowy.

Czy pozostałe braki są zależne od zmiennej `survived`?

```python
df[cols_with_nan].isnull().groupby(df["survived"]).mean()
```

Kolejną kolumną typu **MNAR** okazuje się `boat`: ofiary w większości nie dostały się na żadną z szalup ratunkowych, co było spowodowane niedostateczną liczbą szalup na pokładzie.

### 3.2. Zależność braków od klasy pasażera (`pclass`)

```python
df[cols_with_nan].isnull().groupby(df["pclass"]).mean()
```

```python
# Odsetek ocalałych w podziale na klasę pasażera
df["survived"].astype(int).groupby(df["pclass"]).mean()
```

Braki w kolumnach `cabin` i `home.dest` zależą od zmiennych obserwowanych, więc nie są MCAR:

- klasa pasażera (`pclass`): braki `cabin` to 20,7% w 1. klasie wobec 91,7% w 2. i 97,7% w 3.; braki `home.dest` to odpowiednio 10,5% i 5,8% wobec 72,5%,
- przeżycie (`survived`): braki `cabin` to 87,4% u ofiar wobec 61,4% u ocalałych; braki `home.dest` to 50,8% wobec 30,6%.

**`cabin`:** w zachowanych listach pasażerów numery kabin nie były wpisywane, a pełne plany zakwaterowania zatonęły wraz ze statkiem. Dane pochodzą więc głównie ze wspomnień ocalałych oraz częściowej listy stewarda dla 1. klasy. Część osób mogła też nie pamiętać numeru. Dodatkowo w 3. klasie kabiny nie były przydzielane przed wypłynięciem, tylko na pokładzie.

**`home.dest`:** na listach pasażerów z przesiadkami rejestrowano port wyokrętowania Titanica, a nie adres docelowy w Ameryce. Cel podróży udawało się ustalić głównie dla ocalałych oraz dla dobrze znanych pasażerów 1. i 2. klasy, np. na podstawie przekazów bliskich.

Ponieważ prawdopodobieństwo braku zależy od zmiennych obserwowanych (`pclass`, `survived`), klasyfikuję oba przypadki jako **MAR**.

### 3.3. Klasyfikacja braków danych

1. **MNAR (Missing Not at Random):**
   * **`body`:** 100% braków u ocalałych i 85,0% u ofiar. Brak wynika z tego, że pasażer przeżył lub że jego ciała nie wyłowiono z oceanu.
   * **`boat`:** 98,9% braków u ofiar wobec 4,6% u ocalałych. Brak numeru szalupy bezpośrednio oznacza, że pasażer nie znalazł się w łodzi ratunkowej.
2. **MAR (Missing at Random):**
   * **`cabin`:** Odsetek braków silnie skorelowany z klasą biletu (`pclass` = 3 to ponad 95% braków kabin).
   * **`home.dest`:** 72,5% braków w 3. klasie wobec 6–11% w 1. i 2. klasie. Mechanizm wynika z procedur rejestracji emigrantów.
   * **`age`:** W 3. klasie częściej brakowało dokumentów tożsamości niż w klasach wyższych.
3. **MCAR (Missing Completely at Random):**
   * **`fare`** (1 brak) oraz **`embarked`** (2 braki): Pojedyncze, prawdopodobnie przypadkowe pominięcia przy rejestracji. Oba braki w `embarked` dotyczą pasażerek z jednego biletu (113572, kabina B28), co wskazuje na jeden incydent, a nie dwa niezależne. Przy tak małej liczbie braków (0,15% i 0,08%) można przyjąć MCAR. Wpływ na dalszą analizę jest pomijalny.

## 4. Postępowanie z brakującymi wartościami

* **`boat`, `body` → usunięcie kolumn (ochrona przed *data leakage*):**
  Informacje te powstały w trakcie lub po katastrofie. Użycie ich w modelu predykcyjnym sztucznie zawyżyłoby skuteczność predykcji.
* **`cabin` → inżynieria cech (*feature engineering*):**
  Ze względu na 77,5% braków imputacja konkretnego numeru nie ma sensu. Dobrym rozwiązaniem jest utworzenie flagi binarnej `has_cabin` (0/1) lub skrócenie etykiet zawierających tylko pierwszą literę.
* **`age` → imputacja warunkowa:**
  Kluczowa zmienna biologiczna (20% braków). Usunięcie wierszy spowodowałoby utratę cennych danych. Rekomendowana jest imputacja medianą w podgrupach (wyznaczonych przez `pclass` i tytuł wyciągnięty z nazwiska, np. *Mr*, *Miss*, *Master*) lub algorytmem `KNNImputer`.
* **`embarked`, `fare` → prosta imputacja:**
  * `embarked`: Imputacja dominantą (najczęstszy port: `'S'`).
  * `fare`: Imputacja medianą dla pasażerów 3. klasy.
* **`home.dest` → usunięcie cechy:**
  Wysoka kardynalność i ponad 43% braków – cecha o małej przydatności w standardowej klasyfikacji.
